```python
# import quries 
import pandas as pd
import numpy as np
# data loading 
df = pd.read_csv(r"C:\Users\pc\Downloads\ecommerce_1000_clean.csv")
```

```python
# top 5 row
df.head()
```

```python
# shape data
df.shape
```

```python
# data info
df.info()
```

```python
# find null values
df.isnull().sum()
```

```python
# fill null values in customer Age
df["Customer_Age"]=df["Customer_Age"].fillna(df["Customer_Age"]).mean()
# fill null values in Total_orders
df["Total_Orders"]=df["Total_Orders"].fillna(df["Total_Orders"]).mean()
```

```python
# see data
df.isnull().sum()
```

```python
# find duplicate value
print("duplicate_value:",df.duplicated().sum())
```

```python
# drop duplicate value
df=df.drop_duplicates()
```

```python
# see data of duplicated
print("duplicates_value:",df.duplicated().sum())
```

```python
# new colume add
# convert data type
df["Account_Creation_Date"]=pd.to_datetime(df["Account_Creation_Date"])
df["Month"]=df["Account_Creation_Date"].dt.month_name()
```
